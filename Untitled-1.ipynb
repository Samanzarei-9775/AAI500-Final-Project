```python
# pyright: reportMissingImports=false
# pylint: disable=import-error
# %%

import pandas as pd  
import numpy as np  
from scipy.stats import ttest_ind, t, sem 

# %%
# Load the Houses dataset directly from the textbook's website
url = "http://stat4ds.rwth-aachen.de/data/Houses.dat"
houses = pd.read_table(url, sep=r'\s+')

# Defining the groups of old and new homes
new_homes = houses[houses['new'] == 1]['price']
old_homes = houses[houses['new'] == 0]['price']

# --- Descriptive Statistics ---
n_new = len(new_homes)
n_old = len(old_homes)
mean_new = np.mean(new_homes)
mean_old = np.mean(old_homes)
std_new, std_old = np.std(new_homes, ddof=1), np.std(old_homes, ddof=1)

print("--- DESCRIPTIVE STATISTICS ---")
print(f"New Homes  -> Count: {n_new}, Mean: ${mean_new:.2f}k, Std: {std_new:.2f}")
print(f"Older Homes -> Count: {n_old}, Mean: ${mean_old:.2f}k, Std: {std_old:.2f}")

# %% Second Cell
t_pooled, p_pooled = ttest_ind(new_homes, old_homes, equal_var=True)
var1, var2 = std_new**2, std_old**2
pooled_var = ((n_new - 1) * var1 + (n_old - 1) * var2) / (n_new + n_old - 2)
se_pooled = np.sqrt(pooled_var * (1/n_new + 1/n_old))
df_pooled = n_new + n_old - 2
mean_diff = mean_new - mean_old
ci_pooled = t.interval(0.95, df=df_pooled, loc=mean_diff, scale=se_pooled)

print("\n--- METHOD 1: Pooled-Variance (var.equal = TRUE) ---")

print(f"95% CI for Difference: [${ci_pooled[0]:.2f}k, ${ci_pooled[1]:.2f}k]")

# --- METHOD 2: Welch's Method (Unequal Variances, equal_var=False) ---
t_welch, p_welch = ttest_ind(new_homes, old_homes, equal_var=False)
se_welch = np.sqrt((sem(new_homes)**2) + (sem(old_homes)**2))
df_welch = (se_welch**4) / (((sem(new_homes)**4)/(n_new-1)) + ((sem(old_homes)**4)/(n_old-1)))
ci_welch = t.interval(0.95, df=df_welch, loc=mean_diff, scale=se_welch)
 
print("\n--- METHOD 2: Welch's Method (var.equal = FALSE) ---")
print(f"95% CI for Difference: [${ci_welch[0]:.2f}k, ${ci_welch[1]:.2f}k]")
# %%
```
