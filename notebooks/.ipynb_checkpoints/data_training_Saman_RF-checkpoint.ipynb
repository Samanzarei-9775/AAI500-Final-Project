# Model Training & Evaluation: Random Forest

**Team Member:** Saman

This notebook trains the data by Random Forest on the fully cleaned and standardized dataset produced in `data_cleaning_merged.ipynb`. It evaluates the model strictly on the Validation set to avoid data leakage before final testing.

**Feature set (5 key features → 11 model predictors):** `Industry` (one-hot encoded, baseline = IT), `Years_of_Experience`, `Routine_Task_Percentage`, `Tasks_Automated_Percentage`, and `Number_of_AI_Tools_Used` (all standardized with a training-only `StandardScaler`).

**Target:** `Layoff_Risk` (0 = Low, 1 = Medium, 2 = High).

```python
# Import necessary libraries for data manipulation and mathematical operations
import pandas as pd
import numpy as np
# Import libraries for data visualization
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score
# Configure visual aesthetics for professional graphing
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.titlesize'] = 14
plt.rcParams['axes.titlesize'] = 12
plt.rcParams['axes.labelsize'] = 11
```

## 1. Data Loading
We load the strictly partitioned subsets from the `Data/` directory. For this iteration, we only use `X_train` / `y_train` for model fitting, and `X_val` / `y_val` for intermediate evaluation.

```python
# -------------------------------------------------------------------------
# Load Processed Training and Validation Sets
# -------------------------------------------------------------------------
# Define the relative path to the preprocessed data directory
# Resolve the Data directory dynamically so the notebook runs whether the working
# directory is the project root or the notebooks/ folder (same logic as data_cleaning_merged)
from pathlib import Path
data_dir = Path.cwd() / "Data"
if not data_dir.exists():
    data_dir = (Path.cwd() / ".." / "Data").resolve()
base_path = f"{data_dir.as_posix()}/"

# Load the predictor features (X) for the training subset
X_train = pd.read_csv(f"{base_path}processed_X_train.csv")

# Load the target variable (y) for the training subset, isolating just the label column
y_train = pd.read_csv(f"{base_path}processed_y_train.csv")['Layoff_Risk']

# Load the predictor features (X) for the validation subset
X_val = pd.read_csv(f"{base_path}processed_X_val.csv")

# Load the target variable (y) for the validation subset, isolating just the label column
y_val = pd.read_csv(f"{base_path}processed_y_val.csv")['Layoff_Risk']

# Print the dimensionality to verify successful loading
print(f"Training set: {X_train.shape[0]:,} rows")
print(f"Validation set: {X_val.shape[0]:,} rows")
```

```python
# -------------------------------------------------------------------------
# Dataset Verification: Feature Set & Class Balance
# Confirms that the newly engineered 5-feature pipeline (11 predictors after
# one-hot encoding) was loaded correctly and establishes the majority-class baseline.
# -------------------------------------------------------------------------
print("--- PREDICTOR COLUMNS LOADED (X_train) ---")
for idx, col in enumerate(X_train.columns, start=1):
    print(f"{idx:>2}. {col}")
print(f"\nTotal predictors: {X_train.shape[1]}")

# Confirm train and validation partitions share an identical schema
assert list(X_train.columns) == list(X_val.columns), "Schema mismatch between train and validation sets!"
print("Schema check passed: train and validation columns are identical.")

# Verify there are no missing values that would break model fitting
print(f"Missing values in X_train: {int(X_train.isnull().sum().sum())} | X_val: {int(X_val.isnull().sum().sum())}")

# Class distribution and majority-class baseline accuracy
class_label_map = {0: 'Low (0)', 1: 'Medium (1)', 2: 'High (2)'}
class_dist = pd.DataFrame({
    'Train Count': y_train.value_counts().sort_index(),
    'Train (%)': (y_train.value_counts(normalize=True).sort_index() * 100).round(2),
    'Val Count': y_val.value_counts().sort_index(),
    'Val (%)': (y_val.value_counts(normalize=True).sort_index() * 100).round(2),
}).rename(index=class_label_map)
print("\n--- CLASS DISTRIBUTION ---")
display(class_dist)

# A model that always predicts the most frequent training class sets the floor any real model must beat
majority_class = y_train.value_counts().idxmax()
baseline_acc = (y_val == majority_class).mean()
print(f"Majority-class baseline accuracy on validation: {baseline_acc:.4f}")
```

## 2. Traning Data by Random Forest

### Step 1: Model Initialization and Training (Fitting)

**What We Did:** We initialized and fitted an ensemble `RandomForestClassifier` on the training partition ($N = 14,000$, 70%) using the revised 11-predictor specification

**Hyperparameter Configuration:** We parameterized the ensemble with `n_estimators=100` trees and `random_state=42` to guarantee exact replication across runs.

```python
# =========================================================================
# INITIALIZE AND FIT RANDOM FOREST CLASSIFIER
# =========================================================================

# Instantiate the Random Forest ensemble
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Fit directly on the training partition
rf_model.fit(X_train, y_train)

print("Random Forest Classifier fitted successfully!")
print("Model classes learned:", rf_model.classes_)
print(f"Number of decision trees constructed: {len(rf_model.estimators_)}")
print(f"Feature count evaluated at splits: {rf_model.n_features_in_}")
```

### Step 2: Generating Predictions on the Validation Set (X_val)

**What We Did:** We evaluated our fitted Random Forest ensemble on the 3,000 unseen validation instances ($X_{\text{val}}$) by generating both hard class predictions and continuous probability distributions.

**Inference Mechanism:** Final discrete labels are assigned using majority consensus across the 100 decision trees, while continuous class probabilities represent the fraction of tree votes allocated to each risk tier $[P(\text{Low}), P(\text{Medium}), P(\text{High})]$.

```python
# =========================================================================
# GENERATE PREDICTIONS ON VALIDATION SET (RANDOM FOREST)
# =========================================================================

# Predict discrete class labels via ensemble majority voting
y_val_pred_rf = rf_model.predict(X_val)

# Predict class probability distributions (vote share across 100 trees)
y_val_proba_rf = rf_model.predict_proba(X_val)

# Sanity check: Inspect the first 5 validation predictions
comparison_sample_rf = pd.DataFrame({
    'True_Label': y_val.values[:5],
    'Predicted_Label': y_val_pred_rf[:5],
    'P(Low=0)': y_val_proba_rf[:5, 0].round(3),
    'P(Medium=1)': y_val_proba_rf[:5, 1].round(3),
    'P(High=2)': y_val_proba_rf[:5, 2].round(3)
})

print("--- RANDOM FOREST: FIRST 5 VALIDATION PREDICTIONS ---")
display(comparison_sample_rf)
```

### Step 3: Computing Validation Performance Metrics (Random Forest)

```python
# =========================================================================
# COMPUTE EVALUATION METRICS ON VALIDATION SET (RANDOM FOREST)
# =========================================================================

# -------------------------------------------------------------------------
# Scalar Metrics (Accuracy and Multi-Class ROC-AUC)
# -------------------------------------------------------------------------
val_acc_rf = accuracy_score(y_val, y_val_pred_rf)
val_auc_rf = roc_auc_score(y_val, y_val_proba_rf, multi_class='ovr', average='macro')

print("=== RANDOM FOREST: VALIDATION PERFORMANCE METRICS ===")
print(f"Validation Accuracy: {val_acc_rf:.4f} ({val_acc_rf * 100:.2f}%)")
print(f"Validation ROC-AUC (OvR Macro): {val_auc_rf:.4f}")

# -------------------------------------------------------------------------
# Detailed Multi-Class Classification Report
# -------------------------------------------------------------------------
print("\n--- CLASSIFICATION REPORT (PRECISION, RECALL, F1-SCORE) ---")
print(classification_report(y_val, y_val_pred_rf, target_names=['Low (0)', 'Medium (1)', 'High (2)'], digits=4))

# -------------------------------------------------------------------------
# Confusion Matrix Visualization
# -------------------------------------------------------------------------
cm_rf = confusion_matrix(y_val, y_val_pred_rf)
cm_labels = ['Low', 'Medium', 'High']

fig, ax = plt.subplots(figsize=(6, 4.5))
sns.heatmap(
    cm_rf, 
    annot=True, 
    fmt='d', 
    cmap='Blues', 
    xticklabels=cm_labels, 
    yticklabels=cm_labels,
    cbar=False,
    ax=ax
)
ax.set_title("Validation Confusion Matrix — Random Forest", fontsize=11, fontweight='bold', pad=10)
ax.set_xlabel("Predicted Label", fontsize=10, fontweight='bold')
ax.set_ylabel("True Label", fontsize=10, fontweight='bold')
plt.tight_layout()
plt.show()
```

Evaluating the Random Forest ensemble on the 3,000 unseen validation instances demonstrated strong non-linear classification capabilities, achieving an overall validation accuracy of 72.47% (2,174 out of 3,000 correct) and a One-vs-Rest Macro ROC-AUC of 0.8822. This performance more than doubles the 34.00% majority-class baseline, verifying that the 100-tree bagging ensemble effectively captures structural risk patterns from workload automation and industry features. Per-class metrics indicate solid performance at the risk extremes: Low Risk achieved a precision of 79.52% and recall of 77.68% (F1-score of 0.7859), while High Risk reached a precision of 77.92% and recall of 81.67% (F1-score of 0.7975). Medium Risk yielded an intermediate F1-score of 0.5855 (precision 59.34%, recall 57.78%), demonstrating that boundary ambiguities between adjacent automation percentages remain the primary source of classification variance. The 3 × 3 confusion matrix confirms robust structural separation across polar categories: out of 1,020 actual High-risk workers, 833 were correctly flagged, 182 were estimated as Medium risk, and only 5 were misclassified as Low risk. Similarly, among 990 true Low-risk instances, 769 were correctly identified, 210 were placed in Medium risk, and only 11 (1.1%) were incorrectly escalated to High risk. Out of 990 actual Medium-risk employees, 572 were correctly classified, while 193 were predicted as Low and 225 as High. The concentration of errors almost exclusively along adjacent boundaries rather than opposite extremes demonstrates high stability and calibrated decision-making across the risk continuum.

### Step 4: Gini Feature Importance Extraction and Visualization

```python
# =========================================================================
# EXTRACTING AND VISUALIZING FEATURE IMPORTANCE (RANDOM FOREST)
# =========================================================================

# Extract Gini importances from the fitted Random Forest ensemble
importances = rf_model.feature_importances_
feature_names = X_train.columns

# 2. Build a structured DataFrame with explicit .copy() to eliminate warnings
fi_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances,
    'Importance_Pct': (importances * 100).round(2)
}).sort_values(by='Importance', ascending=False).reset_index(drop=True).copy()

# Compute cumulative contribution cleanly on the explicit copy
fi_df['Cumulative_Pct'] = fi_df['Importance_Pct'].cumsum().round(2)

print("=== RANDOM FOREST: FEATURE IMPORTANCE SUMMARY TABLE ===")
display(fi_df)

# 3. Horizontal Bar Chart of Feature Importances
fig, ax = plt.subplots(figsize=(10, 6))

# Sort ascending for clean bottom-to-top horizontal bar display
plot_fi = fi_df.sort_values(by='Importance', ascending=True)

bars = ax.barh(
    plot_fi['Feature'], 
    plot_fi['Importance_Pct'], 
    color='#3498db', 
    edgecolor='black', 
    linewidth=0.8
)

# Overlay percentage labels on the bars
for bar in bars:
    width = bar.get_width()
    ax.text(
        width + 0.5, 
        bar.get_y() + bar.get_height() / 2, 
        f'{width:.2f}%', 
        va='center', 
        ha='left', 
        fontsize=9.5, 
        fontweight='bold'
    )

ax.set_title("Random Forest: Gini Feature Importance Distribution (%)\n[Mean Decrease in Impurity across 100 Trees]", fontsize=12, fontweight='bold', pad=12)
ax.set_xlabel("Relative Importance (%)", fontsize=11, fontweight='bold')
ax.set_ylabel("Predictor Feature", fontsize=11, fontweight='bold')
ax.set_xlim(0, max(plot_fi['Importance_Pct']) + 5)

plt.tight_layout()
plt.show()
```

Analyzing Gini feature importance across the 100-tree Random Forest ensemble reveals that structural displacement risk is overwhelmingly driven by workload composition and automation exposure rather than industry sector indicators. The top two predictors—Routine Task Percentage (34.04%) and Tasks Automated Percentage (30.14%)—collectively account for 64.18% of all Gini impurity reduction across the forest, serving as the primary split variables in separating layoff tiers. Demographic tenure (Years of Experience) ranks third at 14.77%, followed closely by technological exposure (Number of AI Tools Used) at 11.36%. Together, these four continuous metrics constitute 90.31% of the model's total predictive power. In contrast, all seven one-hot industry indicators contribute modestly, each accounting for between 0.91% and 2.06% of total tree split decisions, with Education (2.06%), Healthcare (1.78%), and Manufacturing (1.52%) offering the strongest sectoral signals. These results indicate that while macro-level industry context plays a measurable baseline role, the ensemble's branching logic relies primarily on an employee's day-to-day routine share and task automation susceptibility to determine individual layoff vulnerability.
